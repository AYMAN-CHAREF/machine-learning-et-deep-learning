#Clustring de BD avec Kmeans

```python
from sklearn.datasets import make_circles
X,y = make_circles(n_samples=150, factor=0.5, noise=0.05,random_state=170)
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
plt.xlabel('x1')
plt.ylabel('x2')
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
jws=[]
scores_sils=[]
for k in range(2,15):
  model=KMeans(n_clusters=k)
  model.fit(X)
  labels=model.labels_
  jws.append(model.inertia_)
  scores_sils.append(silhouette_score(X,labels))
  #print("k=",k,"Jw",model.inertia_,"score=",silhouette_score(X,labels))
```

```python

fig, axes=plt.subplots(1,2,figsize=(10,5))
# Plot inertia values (Jw)
axes[0].plot(np.arange(2, 15), jws, c='red', marker='o')
axes[0].set_xlabel("k")
axes[0].set_ylabel("Jw")
axes[0].set_title("Evolution de Jws en fonction de k")

# Plot silhouette scores
axes[1].plot(np.arange(2, 15), scores_sils, c='blue', marker='o')
axes[1].set_xlabel("k")
axes[1].set_ylabel("score silhouette")
axes[1].set_title("Evolution de score silhouette en fonction de k")
```

```python
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
link=linkage(X,method='ward')
dendrogram(link)
plt.show()
```

```python
model = AgglomerativeClustering(n_clusters=2, linkage="single")
model.fit(X)
labels = model.labels_

plt.scatter(X[:,0],X[:,1],c=labels)
plt.xlabel('x1')
plt.ylabel('x2')
plt.show()
```
