```python
import matplotlib.pyplot as plt
from sklearn import datasets
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
```

```python
n= 1500
noisy_circles = datasets.make_circles(n_samples=n, factor=0.5, noise=0.05, random_state=170)
noisy_moons = datasets.make_moons(n_samples=n, noise=0.05, random_state=170)
blobs = datasets.make_blobs(n_samples=n, random_state=170)
rng = np.random.RandomState(170)
no_structure = rng.rand(n, 2), None

#Anistropicly distrubuted data
X,y = datasets.make_blobs(n_samples=n, random_state=170)
trasformation = [[0,6, -0,6],[-0,4, 0,8]]
X_aniso = np.dot(X, trasformation)
aniso = (X_aniso, y)

#blobs with varied variances
varied = datasets.make_blobs(n_samples=n, cluster_std=[1.0, 2.5, 0.5], random_state=170)

datasets = [noisy_circles, noisy_moons, varied, aniso, blobs, no_structure]

varied
```

```python
colors = ['red', 'blue', 'green', 'purple', 'orange', 'cyan', 'magenta', 'brown', 'yellow','pink', 'Lime', 'Teal', 'Olive', 'Navy', 'Maroon']
```

```python
fig, axes = plt.subplots(1, len(datasets), figsize=(25, 5))

for i in range(len(datasets)):
    X, y = datasets[i]

    for j in range(len(X)):
        if y is not None:

            color = colors[y[j]]
        else:
            color = colors[0]

        # axes[i].scatter(X[j, 0], X[j, 1], c=color)
        axes[i].scatter(X[j, 0], X[j, 1], c=color, marker='o', s=125, alpha=0.1, edgecolor='k')

plt.show()
```

```python
fig, axes = plt.subplots(1, len(datasets), figsize=(25, 5))
k = 2
for i in range(len(datasets)):
  X, y = datasets[i]


  model= KMeans(n_clusters=k)
  model.fit(X)
  labels = model.predict(X)
  Jw= model.inertia_

  axes[i].scatter(X[ :,0], X[ :,1], c=labels, marker='o', s=125, alpha=0.1, edgecolor='k')
  axes[i].set_title("k="+str(k)+"Jw"+str(np.round(Jw,3)), fontsize = 15, c ='b')
  axes[i].set_xlabel("Feature 1")
  axes[i].set_ylabel("Feature 2")

  """axes[i].set_title(f"Dataset {i+1}")
  axes[i].set_xlabel("Feature 1")
  axes[i].set_ylabel("Feature 2")"""

plt.show()
```
