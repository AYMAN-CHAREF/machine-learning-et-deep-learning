```python
from sklearn import datasets

iris = datasets.load_iris()

```

```python
iris
```

```python
#X,Y=iris.data,iris.target
X,Y=iris['data'], iris['target']
```

```python
import matplotlib.pyplot as plt
plt.scatter(X[:,0],X[:,1],c=Y)
plt.show()
```

```python
plt.scatter(X[:,0],X[:,2],c=Y)
plt.show()
```

```python
from sklearn.cluster import KMeans

```

```python
K_clusters=11
listJw =[]
for i in range(2,K_clusters):
  KAlgo=KMeans(n_clusters=i)
  KAlgo.fit(X)
  listJw.append(KAlgo.inertia_)

plt.plot(range(2,K_clusters),listJw)
plt.xlabel('k')
plt.ylabel('Jw')
plt.show()
```

```python
Y_true=Y
KAlgo=KMeans(n_clusters=3)
KAlgo.fit(X)
Y_pred=KAlgo.predict(X)
```
