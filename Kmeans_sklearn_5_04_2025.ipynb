```python
#import des librairies l'environnement

import numpy as np
import sklearn.metrics as sm
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn import datasets
```

```python
#chargement de base de données iris
iris = datasets.load_iris()
```

```python
#affichage des données, vous permet de mieux comprendre le jeu de données (optionnel)
print(iris)
print(iris.data)
print(iris.feature_names)
print(iris.target)
print(iris.target_names)
```

```python

#X,Y=iris.data,iris.target
X,Y=iris['data'], iris['target']
```

```python

import matplotlib.pyplot as plt
plt.scatter(X[:,0],X[:,1],c=Y)
plt.show()
```

```python
plt.scatter(X[:,0],X[:,2],c=Y)
plt.show()
```

```python

K_clusters=11
listJw =[]
for i in range(2,K_clusters):
  KAlgo=KMeans(n_clusters=i)
  KAlgo.fit(X)
  listJw.append(KAlgo.inertia_)

plt.plot(range(2,K_clusters),listJw)
plt.xlabel('k')
plt.ylabel('Jw')
plt.show()
```

```python

Y_true=Y
KAlgo=KMeans(n_clusters=3)
KAlgo.fit(X)
Y_pred=KAlgo.predict(X)
```
